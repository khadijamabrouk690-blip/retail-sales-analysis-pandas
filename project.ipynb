```python
import matplotlib.pyplot as plt
import seaborn as sns
sns.set_theme(style="whitegrid")
```

```python
df  = pd.read_csv('Data Analysis.csv')
```

```python
df
```

```python
## 1- What is the total sales? 
print('The total sales:',  int(df['Total_Sales'].sum()/ 1000000) , 'M')
```

```python
## 2 What is the count of each product?
df['Product'].value_counts()
```

```python
## 3. How many products were sold each year?
df.groupby('Year')['Product'].count().sort_values(ascending=False)
```

```python
agg = df.groupby('Year')['Product'].count().sort_values(ascending=False)
agg.values
```

```python
# 4. Count of each product
count_product = df['Product'].value_counts()
```

```python
# 5. Top product by total sales
top_product = df.groupby('Product')['Total_Sales'].sum().sort_values(ascending=False)
```

```python
# 6. Lowest product by total sales
lowest_product = df.groupby('Product')['Total_Sales'].sum().sort_values().head(1)
lowest_product
```

```python
### Time Analysis
# 7. Products sold each year
units_per_year = df.groupby('Year')['Units_Sold'].sum()
units_per_year
```

```python
# 8. Total sales change over years
sales_per_year = df.groupby('Year')['Total_Sales'].sum()
plt.figure()
sns.barplot(x=sales_per_year.index, y=sales_per_year.values)
plt.title('Total Sales by Year')
plt.savefig('sales_by_year.png', bbox_inches='tight')
plt.show()
```

```python
# 9. Sales month over month
sales_m = df.groupby(['Year','Month'])['Total_Sales'].sum().reset_index()
sales_m['YM'] = sales_m['Year'].astype(str) + '-' + sales_m['Month'].astype(str).str.zfill(2)
plt.figure(figsize=(12,4))
sns.lineplot(data=sales_m, x='YM', y='Total_Sales', marker='o')
plt.xticks(rotation=45); plt.title('Month-over-Month Sales')
plt.savefig('sales_monthly.png', bbox_inches='tight')
plt.show()
```

```python
### Retailer Analysis
# 10. Retailer with highest total sales
best_retailer_sales = df.groupby('Retailer')['Total_Sales'].sum().sort_values(ascending=False)
best_retailer_sales
```

```python
# 11. Retailer with highest operating profit
best_retailer_profit = df.groupby('Retailer')['Operating_Profit'].sum().sort_values(ascending=False)
best_retailer_profit
```

```python
## Retailer by sales and profit

ret = df.groupby('Retailer')[['Total_Sales','Operating_Profit']].sum().sort_values('Total_Sales', ascending=False)
fig, ax = plt.subplots(1,2, figsize=(12,4), sharey=False)
sns.barplot(x=ret.index, y=ret['Total_Sales'], ax=ax[0]); ax[0].set_title('Retailer Sales'); ax[0].tick_params(axis='x', rotation=20)
sns.barplot(x=ret.index, y=ret['Operating_Profit'], ax=ax[1]); ax[1].set_title('Retailer Profit'); ax[1].tick_params(axis='x', rotation=20)
plt.tight_layout(); plt.savefig('retailer.png', bbox_inches='tight'); plt.show()
```

```python
### Product & Profit
# 12. Most profitable product
profit_by_product = df.groupby('Product')['Operating_Profit'].sum().sort_values(ascending=False)
profit_by_product
```

```python
#Total sales and profit by product
prod = df.groupby('Product')[['Total_Sales','Operating_Profit']].sum().sort_values('Total_Sales', ascending=False)
plt.figure(figsize=(10,4))
sns.barplot(x=prod.index, y=prod['Total_Sales'])
plt.xticks(rotation=15); plt.title('Total Sales by Product')
plt.savefig('product_sales.png', bbox_inches='tight'); plt.show()

plt.figure(figsize=(10,4))
sns.barplot(x=prod.sort_values('Operating_Profit', ascending=False).index,
            y=prod.sort_values('Operating_Profit', ascending=False)['Operating_Profit'])
plt.xticks(rotation=15); plt.title('Most Profitable Product')
plt.savefig('product_profit.png', bbox_inches='tight'); plt.show()
```

```python
# 13. Average price per unit per product
avg_price = df.groupby('Product')['Price_Per_Unit'].mean()
avg_price
```

```python
### Geographic
# 14. Region with highest total sales
sales_by_region = df.groupby('Region')['Total_Sales'].sum().sort_values(ascending=False)
sales_by_region
```

```python
# Region by sales
reg = df.groupby('Region')['Total_Sales'].sum().sort_values(ascending=False)
plt.figure()
sns.barplot(x=reg.index, y=reg.values)
plt.title('Sales by Region')
plt.savefig('region.png', bbox_inches='tight'); plt.show()
```

```python
# 15. City contributes most
sales_by_city = df.groupby('City')['Total_Sales'].sum().sort_values(ascending=False)
sales_by_city
```

```python
#Top 10 cities
city = df.groupby('City')['Total_Sales'].sum().sort_values(ascending=False).head(10)
plt.figure()
sns.barplot(x=city.values, y=city.index, orient='h')
plt.title('Top 10 Cities')
plt.savefig('cities.png', bbox_inches='tight'); plt.show()
```
